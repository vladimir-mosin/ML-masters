```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

# ============================================
# 1. Генерация данных
# ============================================
np.random.seed(42)  # фиксируем случайность, чтобы результаты были воспроизводимы

n = 30  # количество квартир

# Площадь — от 30 до 120 кв.м
area = np.random.uniform(30, 120, n)

# Комнаты — от 1 до 5, с привязкой к площади (корреляция)
# rooms = np.round(1 + (area - 30) / 30  + np.random.normal(0, 0.1, n)).clip(1, 15)

rooms = area + np.random.normal(0, 0.01, n)

# Истинные коэффициенты: площадь — 100, комнаты — 50
true_coef_area = 100.0
true_coef_rooms = 50.0
intercept = 0.0

# Цена = истинная цена + шум
noise = np.random.normal(0, 500, n)  
price = true_coef_area * area + true_coef_rooms * rooms + noise

# price[25] = 150000

# Собираем в DataFrame
df = pd.DataFrame({
    'area': area,
    'rooms': rooms,
    'price': price
})

print("Первые 5 строк данных:")
print(df.head())
print(f"\nКорреляция между area и rooms: {df['area'].corr(df['rooms']):.3f}")

# ============================================
# 2. Разделение на train/test
# ============================================
X = df[['area', 'rooms']]
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# ============================================
# 3. Обычная линейная регрессия
# ============================================
ols = LinearRegression()
ols.fit(X_train, y_train)

y_train_pred_ols = ols.predict(X_train)
y_test_pred_ols = ols.predict(X_test)

print("\n=== LinearRegression ===")
print(f"Коэффициенты: area = {ols.coef_[0]:.2f}, rooms = {ols.coef_[1]:.2f}")
print(f"Свободный член: {ols.intercept_:.2f}")
print(f"R² на train: {r2_score(y_train, y_train_pred_ols):.4f}")
print(f"R² на test:  {r2_score(y_test, y_test_pred_ols):.4f}")

# ============================================
# 4. Ridge-регрессия (alpha = 10)
# ============================================

a=10

ridge = Ridge(alpha=a)
ridge.fit(X_train, y_train)

y_train_pred_ridge = ridge.predict(X_train)
y_test_pred_ridge = ridge.predict(X_test)

print("\n=== Ridge (alpha=10) ===")
print(f"Коэффициенты: area = {ridge.coef_[0]:.2f}, rooms = {ridge.coef_[1]:.2f}")
print(f"Свободный член: {ridge.intercept_:.2f}")
print(f"R² на train: {r2_score(y_train, y_train_pred_ridge):.4f}")
print(f"R² на test:  {r2_score(y_test, y_test_pred_ridge):.4f}")

# ============================================
# 5. Дополнительно: сравнение по разным разбиениям
#    чтобы показать, что OLS прыгает, Ridge стабилен
# ============================================
print("\n=== Проверка устойчивости (10 разных разбиений) ===")
ols_coefs = []
ridge_coefs = []

for seed in range(42, 52):
    X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=seed)

    ols = LinearRegression().fit(X_tr, y_tr)
    ridge = Ridge(alpha=a).fit(X_tr, y_tr)

    ols_coefs.append(ols.coef_)
    ridge_coefs.append(ridge.coef_)

ols_coefs = np.array(ols_coefs)
ridge_coefs = np.array(ridge_coefs)

print("Средние коэффициенты OLS:     area = {:.2f} (±{:.2f}), rooms = {:.2f} (±{:.2f})".format(
    ols_coefs[:,0].mean(), ols_coefs[:,0].std(),
    ols_coefs[:,1].mean(), ols_coefs[:,1].std()))
print("Средние коэффициенты Ridge:   area = {:.2f} (±{:.2f}), rooms = {:.2f} (±{:.2f})".format(
    ridge_coefs[:,0].mean(), ridge_coefs[:,0].std(),
    ridge_coefs[:,1].mean(), ridge_coefs[:,1].std()))
```
