```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning) 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
sns.set_style("whitegrid")

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import RobustScaler

from sklearn.preprocessing import PolynomialFeatures


from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsClassifier

from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.linear_model import ElasticNet
from sklearn.linear_model import BayesianRidge

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score, roc_curve
from sklearn.metrics import confusion_matrix
```

```python
import numpy as np
import pandas as pd

# ============================================
# 1. Задаём параметры датасета
# ============================================
N = 500  # количество резюме (строк)
np.random.seed(42)  # для воспроизводимости

# ============================================
# 2. Генерируем предикторы (25 штук)
# ============================================
data = {}

# ID резюме (уникальный)
data['ID'] = np.arange(N)

# 1. Возраст (лет) — от 18 до 65
data['Age'] = np.random.randint(18, 66, N)

# 5. Опыт работы (годы) — сильно коррелирует с возрастом
# Опыт = (возраст - 18) * 0.8 + шум
data['Experience'] = (data['Age'] - 18) * 0.8 + np.random.normal(0, 2, N)
data['Experience'] = np.clip(data['Experience'], 0, 50).astype(int)

# 2. Пол — бинарный (0 — жен, 1 — муж)
data['Gender'] = np.random.binomial(1, 0.5, N)

# 3. Высшее образование
data['HigherEd'] = np.random.binomial(1, 0.85, N)  # у 85% есть

# 4. Красный диплом (только если есть высшее)
data['RedDiploma'] = np.where(
    data['HigherEd'] == 1,
    np.random.binomial(1, 0.2, N),  # 20% от образованных
    0
)

# 6. Количество смен работы за 5 лет
data['JobChanges'] = np.random.poisson(1.5, N)  # среднее ~1.5

# 7. Сертификаты
data['Certificates'] = np.random.binomial(1, 0.3, N)

# 8-9. Конференции и публикации
data['Conferences'] = np.random.binomial(1, 0.25, N)
data['Publications'] = np.random.poisson(0.5, N)  # мало публикуются

# 10. Английский (0-4)
data['English'] = np.random.choice(np.arange(5), N, p=[0.1, 0.2, 0.3, 0.25, 0.15])

# 11. Опыт за рубежом (коррелирует с английским: кто знает язык — чаще ездят)
p_abroad = 0.05 + 0.1 * (data['English'] / 4)  # от 5% до 15%
data['AbroadExp'] = np.random.binomial(1, p_abroad, N)

# 12-13. Готовность к переезду и командировкам
data['Relocation'] = np.random.binomial(1, 0.4, N)
data['BusinessTrips'] = np.random.binomial(1, 0.6, N)

# 14-15. Права и автомобиль (сильная корреляция)
data['DrivingLicense'] = np.random.binomial(1, 0.7, N)
# Если есть права — 80% имеют машину, если нет — только 10%
p_car = np.where(data['DrivingLicense'] == 1, 0.8, 0.1)
data['Car'] = np.random.binomial(1, p_car, N)

# 16-19. Технические навыки (0-5)
for skill, mean in [('Python', 3.0), ('SQL', 2.8), ('R', 1.5), ('Java', 1.2)]:
    data[skill] = np.clip(np.random.normal(mean, 1.5, N).astype(int), 0, 5)

# Python и SQL коррелируют (специалисты обычно знают оба)
data['SQL'] = data['Python'] + np.random.choice([-1, 0, 1], N, p=[0.2, 0.6, 0.2])
data['SQL'] = np.clip(data['SQL'], 0, 5)

# 20. Проекты на GitHub
data['GitHub'] = np.random.poisson(2, N)

# 21. Рекомендации
data['Recommendations'] = np.random.binomial(1, 0.3, N)

# 22. Средняя продолжительность работы на одном месте
# = опыт / количество смен (аккуратно с делением на 0)
data['AvgTenure'] = np.where(
    data['JobChanges'] > 0,
    data['Experience'] / (data['JobChanges'] + 1),
    data['Experience']
)
data['AvgTenure'] = np.round(data['AvgTenure'], 1)

# 23. Хобби в IT
data['HobbyIT'] = np.random.binomial(1, 0.35, N)

# 24. Профпереподготовка
data['Retraining'] = np.random.binomial(1, 0.2, N)

# 25. Город — категориальный
cities = ['Москва', 'СПб', 'Казань', 'Новосибирск', 'Другой']
data['City'] = np.random.choice(cities, N, p=[0.3, 0.2, 0.15, 0.1, 0.25])

# ============================================
# 3. Задаём "истинные" коэффициенты (по 10-балльной шкале)
# ============================================
coeffs = {
    'Age': 2,
    'Gender': -1,
    'HigherEd': 4,
    'RedDiploma': 1,
    'Experience': 6,
    'JobChanges': -2,
    'Certificates': 2,
    'Conferences': 1,
    'Publications': 1,
    'English': 5,
    'AbroadExp': 3,
    'Relocation': 2,
    'BusinessTrips': 1,
    'DrivingLicense': 0,  # нулевой
    'Car': 0,             # нулевой
    'Python': 8,
    'SQL': 6,
    'R': 2,
    'Java': 1,
    'GitHub': 3,
    'Recommendations': 3,
    'AvgTenure': 2,
    'HobbyIT': 1,
    'Retraining': 1,
    # Город — обработаем отдельно через one-hot и веса: Москва=3, СПб=2, Казань=1, Новосибирск=1, Другой=0
}

# ============================================
# 4. Составляем матрицу X
# ============================================
# Категориальный город -> one-hot (пропустим "Другой" как базовый уровень)
city_dummies = pd.get_dummies(data['City'], prefix='City').drop(columns='City_Другой')

# Собираем DataFrame с числовыми предикторами
df = pd.DataFrame({
    'Age': data['Age'],
    'Gender': data['Gender'],
    'HigherEd': data['HigherEd'],
    'RedDiploma': data['RedDiploma'],
    'Experience': data['Experience'],
    'JobChanges': data['JobChanges'],
    'Certificates': data['Certificates'],
    'Conferences': data['Conferences'],
    'Publications': data['Publications'],
    'English': data['English'],
    'AbroadExp': data['AbroadExp'],
    'Relocation': data['Relocation'],
    'BusinessTrips': data['BusinessTrips'],
    'DrivingLicense': data['DrivingLicense'],
    'Car': data['Car'],
    'Python': data['Python'],
    'SQL': data['SQL'],
    'R': data['R'],
    'Java': data['Java'],
    'GitHub': data['GitHub'],
    'Recommendations': data['Recommendations'],
    'AvgTenure': data['AvgTenure'],
    'HobbyIT': data['HobbyIT'],
    'Retraining': data['Retraining'],
})

# Добавляем one-hot города
df = pd.concat([df, city_dummies], axis=1)

# Словарь коэффициентов теперь дополняем городами
city_coeffs = {'City_Москва': 3, 'City_СПб': 2, 'City_Казань': 1, 'City_Новосибирск': 1}
coeffs.update(city_coeffs)

# ============================================
# 5. Считаем таргет (линейная комбинация + шум)
# ============================================
X = df[list(coeffs.keys())].values
true_coef_array = np.array(list(coeffs.values()))

# Линейная комбинация
y_linear = X @ true_coef_array

# Добавляем шум (нормальный, со стандартным отклонением = 3)
noise = np.random.normal(0, 3, N)
y_continuous = y_linear + noise

# ============================================
# 6. Переводим в бинарную метку (0/1) по порогу
# ============================================
median_y = np.median(y_continuous)
y_binary = (y_continuous > median_y).astype(int)

# ============================================
# 7. Собираем итоговый датасет
# ============================================
final_df = df.copy()
final_df['Target'] = y_binary
final_df.insert(0, 'ID', data['ID'])  # ID первой колонкой
final_df['City'] = data['City']       # исходный город тоже сохраним (для наглядности)

# ============================================
# 8. Сохраняем в CSV (без индексов)
# ============================================
final_df.to_csv('hh_ru_dataset.csv', index=False, encoding='utf-8-sig')

print("Датасет сгенерирован: 'hh_ru_dataset.csv'")
print(f"Форма: {final_df.shape}")
print(f"Баланс классов: {final_df['Target'].value_counts().to_dict()}")
print("\nИстинные коэффициенты (для преподавателя):")
for var, coef in coeffs.items():
    print(f"  {var}: {coef}")
```

```python
df = final_df.copy()

X = df.drop(columns = ['Target', 'City', 'ID'])
y = df['Target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5)

knn_1 = KNeighborsClassifier()
knn_1.fit(X_train, y_train)

y_pred_1 = knn_1.predict(X_test)
y_pred_proba_1 = knn_1.predict_proba(X_test)

knn_2 = KNeighborsClassifier()
knn_2.fit(X_test, y_test)

y_pred_2 = knn_2.predict(X_train)
y_pred_proba_2 = knn_2.predict_proba(X_train)

d_1 = pd.DataFrame()
d_1['y_binary'] = y_test
d_1['y_proba'] = y_pred_proba_1[:,1]


d_2 = pd.DataFrame()
d_2['y_binary'] = y_train
d_2['y_proba'] = y_pred_proba_2[:,1]

d = pd.concat([d_1, d_2])
d['index'] = d.index
d.nunique()
```

```python
df = pd.concat([df, d], axis=1)
df
```

```python
df = df.drop(columns = ['y_binary', 'City', 'ID', 'Target'])
df
```

```python
X = df.drop(columns = ['y_proba'])
y = df['y_proba']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

lr = LinearRegression()
lr.fit(X_train, y_train)

y_train_pred_lr = lr.predict(X_train)
y_test_pred_lr = lr.predict(X_test)

print('Линейная регрессия')
print('R² на train:', round(r2_score(y_train, y_train_pred_lr), 4))
print('R² на test:',  round(r2_score(y_test, y_test_pred_lr), 4))
print('')

ridge = Ridge()
ridge.fit(X_train, y_train)

y_train_pred_ridge = ridge.predict(X_train)
y_test_pred_ridge = ridge.predict(X_test)

print('Ридж')
print('R² на train:', round(r2_score(y_train, y_train_pred_ridge), 4))
print('R² на test:',  round(r2_score(y_test, y_test_pred_ridge), 4))
print('')

lasso = Lasso()
lasso.fit(X_train, y_train)

y_train_pred_lasso = lasso.predict(X_train)
y_test_pred_lasso = lasso.predict(X_test)

print('Лассо')
print('R² на train:', round(r2_score(y_train, y_train_pred_lasso), 4))
print('R² на test:',  round(r2_score(y_test, y_test_pred_lasso), 4))
```

```python
X = df.drop(columns = ['y_proba'])
y = df['y_proba']

coef_lr = pd.DataFrame(columns = X.columns)

for k in range(100):

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=k)
    
    scaler = MinMaxScaler()
    scaler.fit(X_train)

    X_train = scaler.transform(X_train)
    X_test = scaler.transform(X_test)

    lr = LinearRegression()
    lr.fit(X_train, y_train)

    y_train_pred_lr = lr.predict(X_train)
    y_test_pred_lr = lr.predict(X_test)

    coef_lr.loc[k] = lr.coef_

###################################    
coef_lr
###################################

coef_long_lr = pd.DataFrame(columns = ['model', 'feature', 'coef'])
for col in coef_lr.columns:
    d = pd.DataFrame(columns = ['model', 'feature', 'coef'])
    d['model'] = ['OLS']*len(coef_lr)
    d['feature'] = [col]*len(coef_lr)
    d['coef'] = coef_lr[col]
    coef_long_lr = pd.concat([coef_long_lr, d], ignore_index=True)
    
###################################
coef_long_lr
###################################

# plt.figure(figsize=(20, 5))
# sns.boxplot(data=coef_long_lr, x='feature', y='coef')
# plt.xticks(rotation=90)
# plt.show()

###################################

X = df.drop(columns = ['y_proba'])
y = df['y_proba']

coef_ridge = pd.DataFrame(columns = X.columns)

for k in range(100):

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=k)
    
    scaler = MinMaxScaler()
    scaler.fit(X_train)

    X_train = scaler.transform(X_train)
    X_test = scaler.transform(X_test)

    ridge = Ridge()
    ridge.fit(X_train, y_train)

    y_train_pred_ridge = ridge.predict(X_train)
    y_test_pred_ridge = ridge.predict(X_test)

    coef_ridge.loc[k] = ridge.coef_

###################################
coef_ridge
###################################

coef_long_ridge = pd.DataFrame(columns = ['model', 'feature', 'coef'])
for col in coef_ridge.columns:
    d = pd.DataFrame(columns = ['model', 'feature', 'coef'])
    d['model'] = ['Ridge']*len(coef_ridge)
    d['feature'] = [col]*len(coef_ridge)
    d['coef'] = coef_ridge[col]
    coef_long_ridge = pd.concat([coef_long_ridge, d], ignore_index=True)

###################################
coef_long_ridge
###################################

coef_long = pd.concat([coef_long_lr, coef_long_ridge], ignore_index=True)

plt.figure(figsize=(20, 5))
sns.boxplot(data=coef_long, x='feature', y='coef', hue = 'model')
plt.xticks(rotation=90)
plt.show()
```
