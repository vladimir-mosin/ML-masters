```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# ============================================
# 1. Параметры генерации
# ============================================
N = 300  # Чем меньше N, тем сильнее будет разброс OLS
np.random.seed(42)

# ============================================
# 2. Генерация признаков
# ============================================
data = {}

data['Age'] = np.random.randint(21, 60, N)
data['Experience'] = (data['Age'] - 21) * 0.7 + np.random.normal(0, 2, N)
data['Experience'] = np.clip(data['Experience'], 0, 35).astype(int)
data['Gender'] = np.random.binomial(1, 0.48, N)
data['HigherEd'] = np.random.binomial(1, 0.8, N)
data['English'] = np.random.choice(np.arange(6), N, p=[0.05, 0.10, 0.20, 0.30, 0.25, 0.10])
data['Python'] = np.clip(np.random.normal(3.0, 1.5, N).astype(int), 0, 5)
data['SQL'] = data['Python'] + np.random.choice([-1, 0, 1], N, p=[0.15, 0.65, 0.20])
data['SQL'] = np.clip(data['SQL'], 0, 5)
data['R'] = np.clip(np.random.normal(2.0, 1.5, N).astype(int), 0, 5)
data['Conferences'] = np.random.binomial(1, 0.30, N)
data['Publications'] = np.random.poisson(0.7, N)
data['Certificates'] = np.random.binomial(1, 0.35, N)
data['AvgTenure'] = np.where(data['Experience'] > 0,
                              data['Experience'] / (1 + np.random.poisson(1.5, N)),
                              data['Experience'] + 0.5)
data['AvgTenure'] = np.round(data['AvgTenure'], 1)

# ============================================
# 3. МУЛЬТИКОЛЛИНЕАРНЫЙ БЛОК (то, ради чего всё)
# ============================================
# Создаём три почти одинаковых признака — они будут сильно коррелировать
base_comm = np.random.normal(0, 1, N)  # латентный фактор "коммуникабельность"
data['Comm_1'] = base_comm + np.random.normal(0, 0.2, N)  # почти то же самое
data['Comm_2'] = base_comm + np.random.normal(0, 0.2, N)
data['Comm_3'] = base_comm + np.random.normal(0, 0.2, N)
# Нормируем их, чтобы были в разумном диапазоне (от -3 до 3 примерно)
for col in ['Comm_1', 'Comm_2', 'Comm_3']:
    data[col] = (data[col] - data[col].mean()) / data[col].std()

# Ещё одна пара сильно коррелирующих признаков: "Работа в выходные" и "Переработки"
data['WeekendWork'] = np.random.binomial(1, 0.3, N)
data['Overtime'] = data['WeekendWork'] + np.random.binomial(1, 0.15, N)
data['Overtime'] = np.clip(data['Overtime'], 0, 1)

# Обычные слабо коррелирующие признаки
data['Leadership'] = np.random.binomial(1, 0.2, N)
data['StartupExp'] = np.random.binomial(1, 0.15, N)

# Собираем датафрейм
df = pd.DataFrame(data)

# ============================================
# 4. ИСТИННЫЕ КОЭФФИЦИЕНТЫ (ты сможешь их менять!)
# ============================================
true_coeffs = {
    'Age': 0.8,
    'Experience': 1.2,
    'Gender': -0.5,
    'HigherEd': 0.6,
    'English': 1.0,
    'Python': 1.8,
    'SQL': 1.2,
    'R': 0.4,
    'Conferences': 0.3,
    'Publications': 0.2,
    'Certificates': 0.5,
    'AvgTenure': 0.4,
    # Мультиколлинеарные — у всех одинаковый истинный вес
    'Comm_1': 1.0,
    'Comm_2': 1.0,
    'Comm_3': 1.0,
    # Ещё пара
    'WeekendWork': -0.6,
    'Overtime': -0.6,
    # Обычные
    'Leadership': 0.7,
    'StartupExp': 0.9,
}

# ============================================
# 5. ФОРМИРОВАНИЕ ТАРГЕТА (через сигмоиду → непрерывный от 0 до 1)
# ============================================
X = df[list(true_coeffs.keys())].values
true_coef_array = np.array(list(true_coeffs.values()))

# Линейная комбинация
y_linear = X @ true_coef_array

# Добавляем шум в логиты (чтобы не было идеальной детерминированности)
logits = y_linear + np.random.normal(0, 2.0, N)

# Сигмоида: y_proba = 1 / (1 + exp(-logits))
y_proba = 1 / (1 + np.exp(-logits))

# Добавляем таргет в датафрейм
df['y_proba'] = y_proba

print(f"Датасет готов: {df.shape[0]} строк, {len(true_coeffs)} признаков + таргет")
print(f"Таргет: от {y_proba.min():.3f} до {y_proba.max():.3f}, среднее = {y_proba.mean():.3f}")
print("\nИстинные коэффициенты (настраивай их смело):")
for k, v in true_coeffs.items():
    print(f"  {k:15s}: {v:+.2f}")

# ============================================
# 6. ДЕМОНСТРАЦИЯ: OLS vs RIDGE (сравнение)
# ============================================
# Масштабируем признаки (обязательно для Ridge)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Обучаем модели
ols = LinearRegression()
ols.fit(X_scaled, y_proba)

ridge = Ridge(alpha=10.0)  # alpha подбирай под задачу
ridge.fit(X_scaled, y_proba)

# Собираем результаты в таблицу
results = pd.DataFrame({
    'Признак': list(true_coeffs.keys()),
    'Истинный': list(true_coeffs.values()),
    'OLS': ols.coef_.flatten() if len(ols.coef_.shape) > 1 else ols.coef_,
    'Ridge': ridge.coef_.flatten() if len(ridge.coef_.shape) > 1 else ridge.coef_,
})

# Вычисляем MAE — насколько коэффициенты отклоняются от истинных
results['|OLS - Истина|'] = np.abs(results['OLS'] - results['Истинный'])
results['|Ridge - Истина|'] = np.abs(results['Ridge'] - results['Истинный'])

print("\n=== Сравнение коэффициентов (масштабированные признаки) ===")
print(results.to_string(index=False))

print(f"\nСредняя ошибка (MAE) OLS:  {results['|OLS - Истина|'].mean():.3f}")
print(f"Средняя ошибка (MAE) Ridge: {results['|Ridge - Истина|'].mean():.3f}")

# ============================================
# 7. ВИЗУАЛИЗАЦИЯ: ящики с усами для OLS и Ridge
# ============================================
# Чтобы нарисовать ящики, нужно сделать много итераций
n_iter = 200  # можно увеличить до 500, но будет дольше
ols_coefs_list = []
ridge_coefs_list = []

for i in range(n_iter):
    # Каждый раз генерируем новые шумы
    logits_i = y_linear + np.random.normal(0, 2.0, N)
    y_proba_i = 1 / (1 + np.exp(-logits_i))

    ols_i = LinearRegression()
    ols_i.fit(X_scaled, y_proba_i)
    ols_coefs_list.append(ols_i.coef_)

    ridge_i = Ridge(alpha=10.0)
    ridge_i.fit(X_scaled, y_proba_i)
    ridge_coefs_list.append(ridge_i.coef_)

# Превращаем списки в массивы
ols_arr = np.array(ols_coefs_list)  # (n_iter, n_features)
ridge_arr = np.array(ridge_coefs_list)

# Собираем в длинный формат для seaborn
feature_names = list(true_coeffs.keys())
df_plot = pd.DataFrame({
    'Feature': np.tile(feature_names, n_iter * 2),
    'Coefficient': np.concatenate([ols_arr.flatten(), ridge_arr.flatten()]),
    'Model': np.concatenate([
        np.repeat('OLS', n_iter * len(feature_names)),
        np.repeat('Ridge', n_iter * len(feature_names))
    ])
})

plt.figure(figsize=(14, 6))
sns.boxplot(data=df_plot, x='Feature', y='Coefficient', hue='Model')
plt.xticks(rotation=45)
plt.title('Разброс коэффициентов: OLS (синий) vs Ridge (оранжевый)\n'
          '(на мультиколлинеарных признаках разница драматическая!)')
plt.tight_layout()
plt.show()
```

```python
X = df.drop(columns = ['y_proba'])
y = df['y_proba']

coef_lr = pd.DataFrame(columns = X.columns)

for k in range(100):

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=k)
    
    scaler = MinMaxScaler()
    scaler.fit(X_train)

    X_train = scaler.transform(X_train)
    X_test = scaler.transform(X_test)

    lr = LinearRegression()
    lr.fit(X_train, y_train)

    y_train_pred_lr = lr.predict(X_train)
    y_test_pred_lr = lr.predict(X_test)

    coef_lr.loc[k] = lr.coef_

###################################    
coef_lr
###################################

coef_long_lr = pd.DataFrame(columns = ['model', 'feature', 'coef'])
for col in coef_lr.columns:
    d = pd.DataFrame(columns = ['model', 'feature', 'coef'])
    d['model'] = ['OLS']*len(coef_lr)
    d['feature'] = [col]*len(coef_lr)
    d['coef'] = coef_lr[col]
    coef_long_lr = pd.concat([coef_long_lr, d], ignore_index=True)
    
###################################
coef_long_lr
###################################

# plt.figure(figsize=(20, 5))
# sns.boxplot(data=coef_long_lr, x='feature', y='coef')
# plt.xticks(rotation=90)
# plt.show()

###################################

X = df.drop(columns = ['y_proba'])
y = df['y_proba']

coef_ridge = pd.DataFrame(columns = X.columns)

for k in range(100):

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=k)
    
    scaler = MinMaxScaler()
    scaler.fit(X_train)

    X_train = scaler.transform(X_train)
    X_test = scaler.transform(X_test)

    ridge = Ridge()
    ridge.fit(X_train, y_train)

    y_train_pred_ridge = ridge.predict(X_train)
    y_test_pred_ridge = ridge.predict(X_test)

    coef_ridge.loc[k] = ridge.coef_

###################################
coef_ridge
###################################

coef_long_ridge = pd.DataFrame(columns = ['model', 'feature', 'coef'])
for col in coef_ridge.columns:
    d = pd.DataFrame(columns = ['model', 'feature', 'coef'])
    d['model'] = ['Ridge']*len(coef_ridge)
    d['feature'] = [col]*len(coef_ridge)
    d['coef'] = coef_ridge[col]
    coef_long_ridge = pd.concat([coef_long_ridge, d], ignore_index=True)

###################################
coef_long_ridge
###################################

coef_long = pd.concat([coef_long_lr, coef_long_ridge], ignore_index=True)

plt.figure(figsize=(20, 5))
sns.boxplot(data=coef_long, x='feature', y='coef', hue = 'model')
plt.xticks(rotation=90)
plt.show()
```
