```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# ============================================
# 1. Параметры генерации
# ============================================
N = 20  # Чем меньше N, тем сильнее будет разброс OLS
np.random.seed(42)

# ============================================
# 2. Генерация признаков
# ============================================
data = {}

data['Age'] = np.random.randint(21, 60, N)
data['Experience'] = (data['Age'] - 21) * 0.7 + np.random.normal(0, 2, N)
data['Experience'] = np.clip(data['Experience'], 0, 35).astype(int)
data['Gender'] = np.random.binomial(1, 0.48, N)
data['HigherEd'] = np.random.binomial(1, 0.8, N)
data['English'] = np.random.choice(np.arange(6), N, p=[0.05, 0.10, 0.20, 0.30, 0.25, 0.10])
data['Python'] = np.clip(np.random.normal(3.0, 1.5, N).astype(int), 0, 5)
data['SQL'] = data['Python'] + np.random.choice([-1, 0, 1], N, p=[0.15, 0.65, 0.20])
data['SQL'] = np.clip(data['SQL'], 0, 5)
data['R'] = np.clip(np.random.normal(2.0, 1.5, N).astype(int), 0, 5)
data['Conferences'] = np.random.binomial(1, 0.30, N)
data['Publications'] = np.random.poisson(0.7, N)
data['Certificates'] = np.random.binomial(1, 0.35, N)
data['AvgTenure'] = np.where(data['Experience'] > 0,
                              data['Experience'] / (1 + np.random.poisson(1.5, N)),
                              data['Experience'] + 0.5)
data['AvgTenure'] = np.round(data['AvgTenure'], 1)

# ============================================
# 3. МУЛЬТИКОЛЛИНЕАРНЫЙ БЛОК (то, ради чего всё)
# ============================================
# Создаём три почти одинаковых признака — они будут сильно коррелировать
base_comm = 0.2 * data['Age'] + np.random.normal(0, 1, N)  # латентный фактор "коммуникабельность"
data['Comm_1'] = base_comm + np.random.normal(0, 0.2, N)  # почти то же самое
data['Comm_2'] = base_comm + np.random.normal(0, 0.2, N)
data['Comm_3'] = base_comm + np.random.normal(0, 0.2, N)


# base_mid = np.random.normal(0, 1, N)
# data['Mid_1'] = base_mid + np.random.normal(0, 0.5, N)  # шум побольше
# data['Mid_2'] = base_mid + np.random.normal(0, 0.5, N)
# data['Mid_3'] = base_mid + np.random.normal(0, 0.5, N)


# Нормируем их, чтобы были в разумном диапазоне (от -3 до 3 примерно)
for col in ['Comm_1', 'Comm_2', 'Comm_3']:
    data[col] = (data[col] - data[col].mean()) / data[col].std()

# Ещё одна пара сильно коррелирующих признаков: "Работа в выходные" и "Переработки"
data['WeekendWork'] = np.random.binomial(1, 0.3, N)
data['Overtime'] = data['WeekendWork'] + np.random.binomial(1, 0.15, N)
data['Overtime'] = np.clip(data['Overtime'], 0, 1)

# Обычные слабо коррелирующие признаки
data['Leadership'] = np.random.binomial(1, 0.2, N)
data['StartupExp'] = np.random.binomial(1, 0.15, N)

# Собираем датафрейм
df = pd.DataFrame(data)

# ============================================
# 4. ИСТИННЫЕ КОЭФФИЦИЕНТЫ (ты сможешь их менять!)
# ============================================
true_coeffs = {
    'Age': -0.8,
    'Experience': 1.2,
    'Gender': -0.5,
    'HigherEd': 0.6,
    'English': 1.0,
    'Python': 3,
    'SQL': 2,
    'R': 0.4,
    'Conferences': 3,
    'Publications': 2,
    'Certificates': 2,
    'AvgTenure': 0.4,
    # Мультиколлинеарные — у всех одинаковый истинный вес
    'Comm_1': 0.1,
    'Comm_2': 0.1,
    'Comm_3': 0.1,
    # Ещё пара
    'WeekendWork': -0.6,
    'Overtime': -0.6,
    # Обычные
    'Leadership': 0.7,
    'StartupExp': 0.9,
}

# ============================================
# 5. ФОРМИРОВАНИЕ ТАРГЕТА (через сигмоиду → непрерывный от 0 до 1)
# ============================================
X = df[list(true_coeffs.keys())].values
true_coef_array = np.array(list(true_coeffs.values()))

# Линейная комбинация
y_linear = X @ true_coef_array

# Добавляем шум в логиты (чтобы не было идеальной детерминированности)
logits = y_linear + np.random.normal(0, 1.25, N)

# Сигмоида: y_proba = 1 / (1 + exp(-logits))
# y_proba = 1 / (1 + np.exp(-logits))

y_proba = logits

# Нет, Маруська. Минимакс
# y_proba = (logits - logits.min())/(logits.max() - logits.min())

# Добавляем таргет в датафрейм
df['Target'] = y_proba

print(f"Датасет готов: {df.shape[0]} строк, {len(true_coeffs)} признаков + таргет")
print(f"Таргет: от {y_proba.min():.3f} до {y_proba.max():.3f}, среднее = {y_proba.mean():.3f}")
```

```python
X = df.drop(columns = ['Target'])
y = df['Target']

max_r2_test_ridge = 0
max_r2_test_lasso = 0

max_alpha_ridge = 0
max_alpha_lasso = 0

train_lr = []
test_lr = []

train_ridge = []
test_ridge = []

train_lasso = []
test_lasso = []

for alpha in np.arange(0, 10, 0.25):

    r2_train_lr = []
    r2_test_lr =[]

    r2_train_ridge = []
    r2_test_ridge =[]

    r2_train_lasso = []
    r2_test_lasso =[]

    for k in range(100):

        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=k)

        lr = LinearRegression()
        lr.fit(X_train, y_train)

        y_train_pred_lr = lr.predict(X_train)
        y_test_pred_lr = lr.predict(X_test)

        r2_train_lr.append(round(r2_score(y_train, y_train_pred_lr), 4))
        r2_test_lr.append(round(r2_score(y_test, y_test_pred_lr), 4))

        ridge = Ridge(alpha)
        ridge.fit(X_train, y_train)

        y_train_pred_ridge = ridge.predict(X_train)
        y_test_pred_ridge = ridge.predict(X_test)

        r2_train_ridge.append(round(r2_score(y_train, y_train_pred_ridge), 4))
        r2_test_ridge.append(round(r2_score(y_test, y_test_pred_ridge), 4))

        lasso = Lasso(alpha)
        lasso.fit(X_train, y_train)

        y_train_pred_lasso = lasso.predict(X_train)
        y_test_pred_lasso = lasso.predict(X_test)

        r2_train_lasso.append(round(r2_score(y_train, y_train_pred_lasso), 4))
        r2_test_lasso.append(round(r2_score(y_test, y_test_pred_lasso), 4))

    train_lr.append(round(np.array(r2_train_lr).mean(), 4))
    test_lr.append(round(np.array(r2_test_lr).mean(), 4))
    
    train_ridge.append(round(np.array(r2_train_ridge).mean(), 4))
    test_ridge.append(round(np.array(r2_test_ridge).mean(), 4))
    
    train_lasso.append(round(np.array(r2_train_lasso).mean(), 4))
    test_lasso.append(round(np.array(r2_test_lasso).mean(), 4))
    
    if round(np.array(r2_test_ridge).mean(), 4) > max_r2_test_ridge:
        max_r2_test_ridge = round(np.array(r2_test_ridge).mean(), 4)
        max_alpha_ridge = alpha
        
    if round(np.array(r2_test_lasso).mean(), 4) > max_r2_test_lasso:
        max_r2_test_lasso = round(np.array(r2_test_lasso).mean(), 4)
        max_alpha_lasso = alpha
        

#     print('')
#     print('')
#     print('alpha =', alpha)
#     print('')
#     print('Линейная регрессия')
#     print('R² на train:', round(np.array(r2_train_lr).mean(), 4))
#     print('R² на test:',  round(np.array(r2_test_lr).mean(), 4))
#     print('')

#     print('Ридж')
#     print('R² на train:', round(np.array(r2_train_ridge).mean(), 4))
#     print('R² на test:',  round(np.array(r2_test_ridge).mean(), 4))
#     print('')

#     print('Лассо')
#     print('R² на train:', round(np.array(r2_train_lasso).mean(), 4))
#     print('R² на test:',  round(np.array(r2_test_lasso).mean(), 4))

print('Ридж')
print('max_r2_test_ridge =', max_r2_test_ridge)
print('max_alpha_ridge =', max_alpha_ridge)
print('')

print('Лассо')
print('max_r2_test_lasso =', max_r2_test_lasso)
print('max_alpha_lasso =', max_alpha_lasso)
```

```python
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(10, 2.5), sharey=True)

ax[0].plot(np.arange(0, 10, 0.25), train_lr, color = 'C0', label = 'train')
ax[0].plot(np.arange(0, 10, 0.25), test_lr, color = 'C1', label = 'test')
ax[0].set_title('Линейная регрессия')
ax[0].legend()

ax[1].plot(np.arange(0, 10, 0.25), train_ridge, color = 'C0', label = 'train')
ax[1].plot(np.arange(0, 10, 0.25), test_ridge, color = 'C1', label = 'test')
ax[1].set_title('Ридж')
ax[1].legend()

ax[2].plot(np.arange(0, 10, 0.25), train_lasso, color = 'C0', label = 'train')
ax[2].plot(np.arange(0, 10, 0.25), test_lasso, color = 'C1', label = 'test')
ax[2].set_title('Лассо')
ax[2].legend()

plt.tight_layout()
plt.show()
```
